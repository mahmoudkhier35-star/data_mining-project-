```python
# Check Dataset Information
import pandas as pd
import numpy as np
from pathlib import Path

# Find the cleaned dataset
possible_paths = [
    Path("data/cleaned_used_car_dataset.csv"),
    Path("regression/data/cleaned_used_car_dataset.csv"),
    Path("Data-Mining-Project/regression/data/cleaned_used_car_dataset.csv")
]

data_path = next((p for p in possible_paths if p.exists()), None)

if data_path is None:
    raise FileNotFoundError("Cleaned dataset not found. Check the file path.")

# Load dataset
df = pd.read_csv(data_path)

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nFirst 5 Rows:")
display(df.head())
```

```python
import pandas as pd

original_df = pd.read_csv("data/used_car_dataset.csv")
cleaned_df = pd.read_csv("data/cleaned_used_car_dataset.csv")

print("Original Dataset:")
print("Shape:", original_df.shape)
print("Duplicates:", original_df.duplicated().sum())

print("\nCleaned Dataset:")
print("Shape:", cleaned_df.shape)
print("Duplicates:", cleaned_df.duplicated().sum())
```

```python
# Remove Duplicate Rows

df = df.drop_duplicates()

print("Dataset Shape After Removing Duplicates:", df.shape)
print("Remaining Duplicates:", df.duplicated().sum())
```

```python
# Separate Features and Target

X = df.drop(columns=["AskPrice", "PostedDate", "AdditionInfo"])
y = df["AskPrice"]

numerical_features = ["Year", "Age", "kmDriven"]

categorical_features = [
    "Brand",
    "model",
    "Transmission",
    "Owner",
    "FuelType"
]

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Create preprocessing transformer

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)

print("Preprocessing Transformer Created Successfully!")
```

```python
from sklearn.feature_selection import SelectKBest, f_regression

feature_selector = SelectKBest(score_func=f_regression, k=10)

print("Feature Selection Tool Created Successfully!")
```

```python
# Display the preprocessing configuration

print("Numerical Features:", numerical_features)
print("Categorical Features:", categorical_features)

print("\nPreprocessing Steps:")
print("1. OneHotEncoder for categorical features")
print("2. StandardScaler for numerical features")

print("\nFeature Selection:")
print("Method: SelectKBest")
print("Scoring Function: f_regression")
print("Number of Features: 10")
```

```python
# Save preprocessing objects for later use

import joblib

joblib.dump(preprocessor, "car_preprocessor.pkl")
joblib.dump(feature_selector, "car_feature_selector.pkl")

print("Preprocessing objects saved successfully!")
```

```python
from sklearn.pipeline import Pipeline

preprocessing_pipeline = Pipeline([
    ("preprocessing", preprocessor),
    ("feature_selection", feature_selector)
])

print("Preprocessing Pipeline Created Successfully!")
```
